```python
# text preprocessing
text = "FREE MONEY NOW!!!"
print(text)

# converting to lowercase
text = text.lower()
print(text)
```

```python
# removing punctuation
import string
text = text.translate(str.maketrans("", "", string.punctuation))
print(text)
```

```python
# removing extra whitespace
text = "   free   money   now   "
text = " ".join(text.split())
print(text)
```

```python
# creating a resusable function for text preprocessing
def clean_text(text):
    text = text.lower()
    text = text.translate(str.maketrans("", "", string.punctuation))
    text = " ".join(text.split())
    return text
```

```python
# testing the function
sample = "   FREE   MONEY   NOW!!!   "
print("Before", sample)
print("After", clean_text(sample))
```

```python
# apply cleaning to entire dataset
combined_df["clean_text"] = combined_df["text"].apply(clean_text)
combined_df[["text", "clean_text"]].head()
```

```python
# verify that the cleaning worked
combined_df[["text", "clean_text", "label"]].sample(10)
```

```python
(combined_df["clean_text"].str.strip() == "").sum()
```

```python
# removing empty messages
combined_df = combined_df[combined_df["clean_text"].str.strip() != ""].reset_index(drop=True)
```
